```python
import sys
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.backends.backend_qt5agg import FigureCanvasQTAgg as FigureCanvas
from matplotlib.figure import Figure
from PyQt5.QtWidgets import *
from PyQt5.QtCore import *
from PyQt5.QtGui import *
import sympy as sp
import re

class Function3DPlotter(QMainWindow):
    def __init__(self):
        super().__init__()
        self.initUI()
        self.x = sp.symbols('x')
        self.y = sp.symbols('y')
        
    def initUI(self):
        self.setWindowTitle("3D Function Plotter - f(x,y)")
        self.setGeometry(100, 100, 1400, 800)
        
        # Central widget and main layout
        central_widget = QWidget()
        self.setCentralWidget(central_widget)
        main_layout = QHBoxLayout(central_widget)
        
        # Left panel for controls
        left_panel = QWidget()
        left_panel.setFixedWidth(400)
        left_layout = QVBoxLayout(left_panel)
        
        # Function input
        function_group = QGroupBox("Function Input")
        function_layout = QVBoxLayout()
        
        self.function_label = QLabel("Enter f(x, y) =")
        function_layout.addWidget(self.function_label)
        
        self.function_input = QLineEdit()
        self.function_input.setPlaceholderText("e.g., sin(x)*cos(y), x**2 + y**2, exp(-x**2-y**2)")
        self.function_input.setText("6 - 3*x - 2*y")
        function_layout.addWidget(self.function_input)
        
        # Examples dropdown
        self.examples_combo = QComboBox()
        self.examples_combo.addItems([
            "Select an example...",
            "Plane: 6 - 3*x - 2*y",
            "Paraboloid: x**2 + y**2",
            "Hyperbolic paraboloid: x**2 - y**2",
            "Sine wave: sin(x) + cos(y)",
            "Gaussian: exp(-x**2 - y**2)",
            "Ripple: sin(x**2 + y**2)",
            "Saddle: x*y",
            "Sphere slice: sqrt(9 - x**2 - y**2)",
            "Logarithmic: log(x**2 + y**2 + 1)"
        ])
        self.examples_combo.currentIndexChanged.connect(self.load_example)
        function_layout.addWidget(QLabel("Examples:"))
        function_layout.addWidget(self.examples_combo)
        
        function_group.setLayout(function_layout)
        left_layout.addWidget(function_group)
        
        # Plot settings
        settings_group = QGroupBox("Plot Settings")
        settings_layout = QGridLayout()
        
        # X range
        settings_layout.addWidget(QLabel("X Min:"), 0, 0)
        self.x_min = QDoubleSpinBox()
        self.x_min.setRange(-50, 50)
        self.x_min.setValue(-3)
        settings_layout.addWidget(self.x_min, 0, 1)
        
        settings_layout.addWidget(QLabel("X Max:"), 0, 2)
        self.x_max = QDoubleSpinBox()
        self.x_max.setRange(-50, 50)
        self.x_max.setValue(3)
        settings_layout.addWidget(self.x_max, 0, 3)
        
        # Y range
        settings_layout.addWidget(QLabel("Y Min:"), 1, 0)
        self.y_min = QDoubleSpinBox()
        self.y_min.setRange(-50, 50)
        self.y_min.setValue(-3)
        settings_layout.addWidget(self.y_min, 1, 1)
        
        settings_layout.addWidget(QLabel("Y Max:"), 1, 2)
        self.y_max = QDoubleSpinBox()
        self.y_max.setRange(-50, 50)
        self.y_max.setValue(3)
        settings_layout.addWidget(self.y_max, 1, 3)
        
        # Resolution
        settings_layout.addWidget(QLabel("Resolution:"), 2, 0)
        self.resolution = QSpinBox()
        self.resolution.setRange(10, 200)
        self.resolution.setValue(50)
        settings_layout.addWidget(self.resolution, 2, 1)
        
        # Color map
        settings_layout.addWidget(QLabel("Color Map:"), 2, 2)
        self.colormap = QComboBox()
        self.colormap.addItems(['viridis', 'plasma', 'inferno', 'magma', 'coolwarm', 'rainbow', 'jet', 'summer', 'winter'])
        settings_layout.addWidget(self.colormap, 2, 3)
        
        # Plot type
        settings_layout.addWidget(QLabel("Plot Type:"), 3, 0)
        self.plot_type = QComboBox()
        self.plot_type.addItems(['Surface', 'Wireframe', 'Contour', 'Surface + Contour'])
        settings_layout.addWidget(self.plot_type, 3, 1)
        
        # Alpha (transparency)
        settings_layout.addWidget(QLabel("Transparency:"), 3, 2)
        self.alpha = QDoubleSpinBox()
        self.alpha.setRange(0.1, 1.0)
        self.alpha.setValue(0.8)
        self.alpha.setSingleStep(0.1)
        settings_layout.addWidget(self.alpha, 3, 3)
        
        settings_group.setLayout(settings_layout)
        left_layout.addWidget(settings_group)
        
        # Evaluation point
        eval_group = QGroupBox("Evaluate at Point")
        eval_layout = QGridLayout()
        
        eval_layout.addWidget(QLabel("x ="), 0, 0)
        self.eval_x = QDoubleSpinBox()
        self.eval_x.setRange(-100, 100)
        self.eval_x.setValue(1.0)
        eval_layout.addWidget(self.eval_x, 0, 1)
        
        eval_layout.addWidget(QLabel("y ="), 0, 2)
        self.eval_y = QDoubleSpinBox()
        self.eval_y.setRange(-100, 100)
        self.eval_y.setValue(1.0)
        eval_layout.addWidget(self.eval_y, 0, 3)
        
        self.eval_button = QPushButton("Evaluate f(x,y)")
        self.eval_button.clicked.connect(self.evaluate_point)
        eval_layout.addWidget(self.eval_button, 1, 0, 1, 2)
        
        self.eval_result = QLabel("f(1.0, 1.0) = ")
        eval_layout.addWidget(self.eval_result, 1, 2, 1, 2)
        
        eval_group.setLayout(eval_layout)
        left_layout.addWidget(eval_group)
        
        # Buttons
        button_layout = QHBoxLayout()
        
        self.plot_button = QPushButton("Plot Function")
        self.plot_button.setStyleSheet("background-color: #4CAF50; color: white; font-weight: bold; padding: 10px;")
        self.plot_button.clicked.connect(self.plot_function)
        button_layout.addWidget(self.plot_button)
        
        self.clear_button = QPushButton("Clear Plot")
        self.clear_button.setStyleSheet("background-color: #f44336; color: white; padding: 10px;")
        self.clear_button.clicked.connect(self.clear_plot)
        button_layout.addWidget(self.clear_button)
        
        left_layout.addLayout(button_layout)
        
        # Status label
        self.status_label = QLabel("Ready to plot. Enter a function and click 'Plot Function'")
        self.status_label.setWordWrap(True)
        self.status_label.setStyleSheet("padding: 10px; background-color: #f0f0f0; border-radius: 5px;")
        left_layout.addWidget(self.status_label)
        
        # Add stretch to push everything to top
        left_layout.addStretch()
        
        # Right panel for plot
        right_panel = QWidget()
        right_layout = QVBoxLayout(right_panel)
        
        # Matplotlib figure
        self.figure = Figure(figsize=(10, 8), dpi=100)
        self.canvas = FigureCanvas(self.figure)
        right_layout.addWidget(self.canvas)
        
        # Add both panels to main layout
        main_layout.addWidget(left_panel)
        main_layout.addWidget(right_panel)
        
        # Apply some styling
        self.setStyleSheet("""
            QMainWindow {
                background-color: #f5f5f5;
            }
            QGroupBox {
                font-weight: bold;
                border: 2px solid #cccccc;
                border-radius: 5px;
                margin-top: 10px;
                padding-top: 10px;
            }
            QGroupBox::title {
                subcontrol-origin: margin;
                left: 10px;
                padding: 0 5px 0 5px;
            }
            QLabel {
                font-size: 12px;
            }
            QPushButton {
                font-size: 14px;
                border-radius: 4px;
                padding: 5px;
            }
        """)
        
        # Plot the default function
        QTimer.singleShot(100, self.plot_function)
        
    def load_example(self, index):
        """Load an example function from the dropdown"""
        if index > 0:  # Skip the first "Select an example..." item
            examples = [
                "6 - 3*x - 2*y",
                "x**2 + y**2",
                "x**2 - y**2",
                "sin(x) + cos(y)",
                "exp(-x**2 - y**2)",
                "sin(x**2 + y**2)",
                "x*y",
                "sqrt(9 - x**2 - y**2)",
                "log(x**2 + y**2 + 1)"
            ]
            self.function_input.setText(examples[index-1])
            self.examples_combo.setCurrentIndex(0)  # Reset dropdown
    
    def evaluate_point(self):
        """Evaluate the function at a specific point"""
        func_str = self.function_input.text().strip()
        if not func_str:
            self.eval_result.setText("Please enter a function first")
            return
            
        try:
            # Convert to sympy expression
            expr = sp.sympify(func_str)
            
            # Evaluate at point
            x_val = self.eval_x.value()
            y_val = self.eval_y.value()
            
            result = expr.subs({self.x: x_val, self.y: y_val})
            result_float = float(result.evalf())
            
            self.eval_result.setText(f"f({x_val}, {y_val}) = {result_float:.4f}")
            self.status_label.setText(f"Evaluation successful: f({x_val}, {y_val}) = {result_float:.4f}")
            
        except Exception as e:
            self.eval_result.setText("Error in evaluation")
            self.status_label.setText(f"Error evaluating function: {str(e)}")
    
    def clear_plot(self):
        """Clear the current plot"""
        self.figure.clear()
        self.canvas.draw()
        self.status_label.setText("Plot cleared. Enter a new function to plot.")
    
    def plot_function(self):
        """Plot the 3D function"""
        func_str = self.function_input.text().strip()
        if not func_str:
            self.status_label.setText("Please enter a function to plot")
            return
        
        try:
            # Clear previous plot
            self.figure.clear()
            
            # Create 3D axis
            ax = self.figure.add_subplot(111, projection='3d')
            
            # Generate grid
            x_min, x_max = self.x_min.value(), self.x_max.value()
            y_min, y_max = self.y_min.value(), self.y_max.value()
            res = self.resolution.value()
            
            x = np.linspace(x_min, x_max, res)
            y = np.linspace(y_min, y_max, res)
            X, Y = np.meshgrid(x, y)
            
            # Convert function string to numpy-compatible format
            # Replace common math functions
            func_np = func_str
            func_np = re.sub(r'\bexp\b', 'np.exp', func_np)
            func_np = re.sub(r'\bsqrt\b', 'np.sqrt', func_np)
            func_np = re.sub(r'\bsin\b', 'np.sin', func_np)
            func_np = re.sub(r'\bcos\b', 'np.cos', func_np)
            func_np = re.sub(r'\btan\b', 'np.tan', func_np)
            func_np = re.sub(r'\blog\b', 'np.log', func_np)
            func_np = re.sub(r'\barcsin\b', 'np.arcsin', func_np)
            func_np = re.sub(r'\barccos\b', 'np.arccos', func_np)
            func_np = re.sub(r'\barctan\b', 'np.arctan', func_np)
            func_np = re.sub(r'\bsinh\b', 'np.sinh', func_np)
            func_np = re.sub(r'\bcosh\b', 'np.cosh', func_np)
            func_np = re.sub(r'\btanh\b', 'np.tanh', func_np)
            
            # Evaluate function
            Z = eval(func_np, {"np": np, "X": X, "Y": Y})
            
            # Handle potential complex results (like sqrt of negative)
            if np.iscomplexobj(Z):
                Z = np.real(Z)
            
            # Plot based on selected type
            plot_type = self.plot_type.currentText()
            cmap = self.colormap.currentText()
            alpha = self.alpha.value()
            
            if plot_type == 'Surface':
                surf = ax.plot_surface(X, Y, Z, cmap=cmap, alpha=alpha, 
                                      linewidth=0.5, antialiased=True)
                self.figure.colorbar(surf, ax=ax, shrink=0.5, aspect=5)
                
            elif plot_type == 'Wireframe':
                ax.plot_wireframe(X, Y, Z, color='blue', alpha=alpha, 
                                 linewidth=0.5)
                
            elif plot_type == 'Contour':
                # Contour plot in 3D
                ax.contour(X, Y, Z, 20, cmap=cmap, offset=np.min(Z))
                surf = ax.plot_surface(X, Y, Z, alpha=0.3, cmap=cmap)
                
            elif plot_type == 'Surface + Contour':
                surf = ax.plot_surface(X, Y, Z, cmap=cmap, alpha=alpha, 
                                      linewidth=0.5, antialiased=True)
                ax.contour(X, Y, Z, 10, colors='k', linewidths=0.5, offset=np.min(Z))
                self.figure.colorbar(surf, ax=ax, shrink=0.5, aspect=5)
            
            # Labels and title
            ax.set_xlabel('X axis', fontsize=12)
            ax.set_ylabel('Y axis', fontsize=12)
            ax.set_zlabel('Z axis', fontsize=12)
            ax.set_title(f'3D Plot of f(x,y) = {func_str}', fontsize=14, pad=20)
            
            # Set axis limits
            ax.set_xlim([x_min, x_max])
            ax.set_ylim([y_min, y_max])
            
            # Grid and view angle
            ax.grid(True, alpha=0.3)
            ax.view_init(elev=30, azim=45)
            
            # Draw the plot
            self.canvas.draw()
            
            # Update status
            self.status_label.setText(f"Successfully plotted: f(x,y) = {func_str}")
            
            # Also evaluate at center point for demonstration
            center_x = (x_min + x_max) / 2
            center_y = (y_min + y_max) / 2
            center_z = eval(func_np, {"np": np, "X": center_x, "Y": center_y})
            self.status_label.setText(f"Plotted: f(x,y) = {func_str}\n"
                                     f"Range: x∈[{x_min},{x_max}], y∈[{y_min},{y_max}]\n"
                                     f"At center ({center_x:.1f},{center_y:.1f}): f≈{float(center_z):.2f}")
            
        except Exception as e:
            self.status_label.setText(f"Error plotting function: {str(e)}")
            # Create empty plot with error message
            self.figure.clear()
            ax = self.figure.add_subplot(111)
            ax.text(0.5, 0.5, f"Error:\n{str(e)}", 
                   ha='center', va='center', fontsize=12, color='red',
                   transform=ax.transAxes)
            ax.set_axis_off()
            self.canvas.draw()

# Main function to run the application
def main():
    app = QApplication(sys.argv)
    app.setStyle('Fusion')  # Modern look
    
    # Set application font
    font = QFont("Segoe UI", 10)
    app.setFont(font)
    
    window = Function3DPlotter()
    window.show()
    
    sys.exit(app.exec_())

if __name__ == '__main__':
    main()
```
